# Physical-field data generation: visualization index

Open one of the case notebooks below and run all cells. Each notebook generates a deterministic periodic reference trajectory and provides an interactive time slider for auditing the proposed distribution, spectral, topological, and PDE-informed losses.

| Case | Target channels | Joint value distribution | Spectrum | Topology view | PDE diagnostics |
| --- | --- | --- | --- | --- | --- |
| [1D chaotic Kuramoto–Sivashinsky](ks/viz.ipynb) | `u` | scalar Wasserstein | 1D power | superlevel components | fourth-order KS residual |
| [1D viscous Burgers](burgers/viz.ipynb) | `u` | scalar Wasserstein | 1D power | superlevel components | advection–diffusion residual |
| [2D Kolmogorov turbulence](navier_stokes/viz.ipynb) | `u, v, p` | joint `(u,v,p)` cloud | kinetic energy | vorticity level sets | momentum, divergence, pressure gauge |

> The level-set panels are topology-oriented previews. They deliberately do not claim to implement the eventual differentiable persistent-homology loss. The earlier [Brusselator notebook](brusselator/viz.ipynb) is retained as an exploratory two-field reaction–diffusion case, but it is not part of the active three-case suite.


## Notebook conventions

- Arrays named `fields` use `[time, channel, x]` in 1D and `[time, channel, y, x]` in 2D.
- Periodic grids exclude the duplicated endpoint.
- All simulations use fixed seeds and display their numerical parameters near the top.
- Animations use Matplotlib's embedded JavaScript controls, so no `ipywidgets` or Plotly installation is required.
- Set the environment variable `PHYCOFLOW_VIZ_SMOKE=1` before execution for the reduced CI/test configuration.
